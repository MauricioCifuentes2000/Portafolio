```python
# Re-run this cell
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
crimes = pd.read_csv("crimes.csv", dtype={"TIME OCC": str})
crimes.head()
```

```python
# Start coding here
# Use as many cells as you need
#extracting the first values
crimes['hour'] = crimes['TIME OCC'].str[:2]
#taking the top counted value from hour column
peak_crime_hour = crimes['hour'].value_counts(sort=True).index[0]
#change type of variable
peak_crime_hour = int(peak_crime_hour)
#print message
print('peak crime hour:',peak_crime_hour)
```

```python
#which area has the largest frequency of night crimes (between 10pm and 3:59am) 
#save string as peak_night_crime_location
#filtering from 22hrs to 03 hrs
night='^22|^23|^24|^00|^01|^02|^03'
#creating a bool columns for the night
crimes['Night'] = crimes['TIME OCC'].str.contains(night)
#creating a new df where Night is true
night_crimes = crimes[crimes['Night'] == True]
#counting areas only for night values and taking the first value
peak_night_crime_location = night_crimes['AREA NAME'].value_counts(sort=True).index[0]
#print message
print('peak_crime_location: ',peak_night_crime_location)
```

```python
#creating a list for the assignment
Victim_ages = ['0-17','18-25','26-34','35-44','45-54','55-64','65+']
#conditions for each group
conditions=[
(crimes['Vict Age'] <= 17),
((crimes['Vict Age'] >= 18) & (crimes['Vict Age'] <= 25)),
((crimes['Vict Age'] >= 26) & (crimes['Vict Age'] <= 34)),
((crimes['Vict Age'] >= 35) & (crimes['Vict Age'] <= 44)),
((crimes['Vict Age'] >= 45) & (crimes['Vict Age'] <= 54)),
((crimes['Vict Age'] >= 55) & (crimes['Vict Age'] <= 64)),
(crimes['Vict Age'] >= 65)
]
#applying the labels from the victim ages list for each condition and creating a column named labels for the result 
crimes['labels'] = np.select(conditions, Victim_ages)
#seting the index using the labels column and saving as new df named crimes labeled
crimes_labeled = crimes.set_index("labels")
#using value counts on the labels column to show the desired result and saving as victim ages
victim_ages = crimes_labeled.value_counts('labels')
```
